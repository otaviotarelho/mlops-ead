```python
os.makedirs('/content/drive/MyDrive/videos', exist_ok=True)
os.makedirs('/content/drive/MyDrive/videos/processed_frames', exist_ok=True)
```

```python
cap = cv2.VideoCapture('/content/drive/MyDrive/videos/IMG_6347.MOV')
```

```python
id_frame = 0
ret = True
while ret:
    ret, frame = cap.read()
    if ret:
        grey = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        cv2.imwrite(f'/content/drive/MyDrive/videos/processed_frames/frame_{id_frame:04d}.jpg', frame)
        id_frame += 1

cap.release()
```
